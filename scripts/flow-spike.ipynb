```python
import os
my_workloads = ["tiled_matmul_ws", "preload_and_compute_sparse_20", "preload_and_compute_sparse_60", "preload_and_compute_sparse_100", "preload_and_compute_sparse_0", "mvin_cache_hit_microbenchmark_0","mvin_cache_hit_microbenchmark_random","mvout_microbenchmark_0","mvout_microbenchmark_random","preload_and_compute_0","preload_and_compute_1","preload_and_compute_random"]
overwrite = True
```

```python
class CommandScript:
    def __init__(self, path):
        self.path = path
        # if cd is not None: print(f"cd {cd}")
        self.f = open(path,'w')
        self.f.write("#!/bin/bash\n")
    def close(self):
        self.f.close()
        os.chmod(self.path, 0o755)
        print("./run-spike.sh")
```

```python
print("cp -R templates/bareMetalC/. ../power-mappings-chipyard/generators/gemmini/software/gemmini-rocc-tests/bareMetalC")
print("cd ../power-mappings-chipyard/generators/gemmini/software/gemmini-rocc-tests")
print("./build.sh")
print("cd ../../../../../scripts")
script = CommandScript(f"run-spike.sh")
for w in my_workloads:
    spike_fpath = f"../data/spike_logs/{w}.spike.out"
    if overwrite or not os.path.exists(spike_fpath):
        script.f.write(f"spike --extension=gemmini --log-commits ../power-mappings-chipyard/generators/gemmini/software/gemmini-rocc-tests/build/bareMetalC/{w}-baremetal &> {spike_fpath}\n")
script.close()
```
